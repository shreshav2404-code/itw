```python
from transformers import pipeline
```

```python
generator=pipeline(
    "text-generation",
    model="openai-community/gpt2",
    token=False,
    device=1
)
```

```python
prompt="how can i be rich "

result=generator(
    prompt,
    max_new_tokens=60,
    do_sample=False,
    pad_token_id=generator.tokenizer.eos_token_id
)

print(result[0]["generated_text"])
```

```python
prompt="how can i be rich "

result=generator(
    prompt,
    max_new_tokens=60,
    do_sample=True,
    temperature=0.1,
    pad_token_id=generator.tokenizer.eos_token_id
)

print(result[0]["generated_text"])
```

Qwen 0.5B Instruct Model

```python
from transformers import pipeline
import torch

# Initialize the pipeline for a small Qwen model
qwen_generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    device="cpu"
)
```

```python
message = "how can i be rich"



response = qwen_generator(
    message,
    max_new_tokens=100,
    do_sample=True,
    temperature=0.5,
    pad_token_id=qwen_generator.tokenizer.eos_token_id
)

print(response[0]["generated_text"][len(prompt):])
```

```python
reviews = [
    "I bought this yesterday and it already broke. Extremely disappointed.",
    "It is decent for the price, but nothing spectacular.",
    "This is the absolute best purchase I have made all year! Highly recommend!"
]

for i, review in enumerate(reviews, 1):
    messages = [
        {
            "role": "system",
            "content": "You are a helpful assistant. Classify the sentiment of the text as either Positive, Negative, or Neutral. Reply with only the classification label."
        },
        {
            "role": "user",
            "content": f"Analyze this text: '{review}'"
        }
    ]

    prompt = qwen_generator.tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

    response = qwen_generator(
        prompt,
        max_new_tokens=10,
        do_sample=False,
        pad_token_id=qwen_generator.tokenizer.eos_token_id
    )

    sentiment = response[0]["generated_text"][len(prompt):].strip()
    print(f"Review {i}: '{review}'")
    print(f"Predicted Sentiment: {sentiment}\n" + "-"*50)
```

```python

qa_messages = [
    {
        "role": "system",
        "content": "You are a helpful assistant. Provide clear, direct, and concise answers to the user's questions in English."
    },
    {
        "role": "user",
        "content": "how can i be rich"
    }
]

qa_prompt = qwen_generator.tokenizer.apply_chat_template(qa_messages, tokenize=False, add_generation_prompt=True)


qa_response = qwen_generator(
    qa_prompt,
    max_new_tokens=150,
    do_sample=True,
    temperature=0.7,
    pad_token_id=qwen_generator.tokenizer.eos_token_id
)


answer = qa_response[0]["generated_text"][len(qa_prompt):].strip()
print(answer)
```
