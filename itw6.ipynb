```python
from sentence_transformers import SentenceTransformers
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
model=sentence_transformers.SentenceTransformer('all-MiniLM-L6-v2')
print("Model loaded")
```

```python
documents = [
    "keshav is an ai engineer",
    "he is from bihar"
    "he likes to eat good food "
    "he earns by himself with stocks , freelancing , gambling and other ways"
    "he is kind hearted and got cheated by his girlfriend"
    "he loves his family "
    "he loves to innovate , learn and create new things"]

query = "is keshav a good person?"
print("Query:", query)


document_embeddings = model.encode(documents)
query_embedding = model.encode([query])
print("Document embeddings shape:", document_embeddings.shape)
print("Query embedding shape:", query_embedding.shape)


```

```python
similarities = cosine_similarity(query_embedding, document_embeddings)
print("Similarities:", similarities)
```
