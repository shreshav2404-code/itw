```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
model=SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2',
token=False
)
```

```python
sentences = [
    "The weather is lovely and sunny today.",
    "It is a beautiful and warm day outside.",
    "Artificial intelligence is transforming industries."
]

embeddings = model.encode(sentences)

print("\nembedding shape :",embeddings.shape)
print("first 10 embeddings values",embeddings[0][:10])

similarities = cosine_similarity(embeddings)


```

```python
print("\n similarity matrix :")
print(similarities.round(3))
```

```python
query="i am feeling cold today"

query_embedding=model.encode([query])


scores=cosine_similarity(query_embedding,embeddings)[0]

best_match=scores.argmax()

print("Query",query)
print("Best Match:",sentences[best_match])
print("Similarity Score:",round(float(scores[best_match]),3))
```

```python

new_sentences = [
    "Deep learning models require high-performance GPUs.",
    "The wind is howling and it's starting to snow.",
    "Machine learning algorithms find patterns in datasets.",
    "I need to grab my winter coat and some gloves.",
    "Global markets reacted positively to the new financial policy."
]


new_embeddings = model.encode(new_sentences)


queries = [
    "Is it going to freeze outside?",
    "How do neural networks learn?",
    "What is the stock market trend?"
]

for query in queries:
    query_emb = model.encode([query])
    scores = cosine_similarity(query_emb, new_embeddings)[0]


    top_3_indices = scores.argsort()[-3:][::-1]

    print(f"\nQuery: '{query}'")

    print("Top 3 matches:")
    for idx in top_3_indices:
        print(f"  - {new_sentences[idx]} (Score: {scores[idx]:.3f})")
```
